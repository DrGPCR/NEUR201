## Step 2 — Variables, math, and text

### 2a. Variables: giving names to values

A **variable** is a labeled container that stores a value so you can reuse it
later. You create one with a single `=` sign: the name goes on the left, the
value on the right.

Think of it like labeling a tube in the lab: once you've written `cell_count`
on the label, you can refer to `cell_count` instead of remembering the number
inside.

```python
cell_count = 42          # store the whole number 42 under the name 'cell_count'
sample_name = "CON1"     # store some text under the name 'sample_name'

print(sample_name)
print(cell_count)
```

Once a value is stored, you can **use** and **update** it. Notice we don't put
quotation marks around a variable name when we want its value.

```python
cell_count = cell_count + 8   # take the current value (42), add 8, store the result back
print(cell_count)             # what number do you expect to see?
```

### 2b. Doing math

Python is a very capable calculator. The basic math symbols are:

| Symbol | Meaning         | Example      |
|:------:|:----------------|:-------------|
| `+`    | add             | `2 + 3` → 5  |
| `-`    | subtract        | `10 - 4` → 6 |
| `*`    | multiply        | `6 * 7` → 42 |
| `/`    | divide          | `20 / 5` → 4 |
| `**`   | power (exponent)| `2 ** 3` → 8 |

Let's compute a **cell density** — cells divided by the imaged area — the exact
kind of calculation you'll do with real data later.

```python
cells = 250          # number of cells counted
area_mm2 = 4.0       # imaged area in square millimeters

density = cells / area_mm2
print(density)       # cells per mm^2
```

> **⚙️ Try it:** Change `cells` to `500` (keep `area_mm2` at `4.0`) and re-run.
> The density should **double**. This is a small preview of why we normalize
> counts to area — it lets us fairly compare images of different sizes.

### 2c. Working with text, and combining text with numbers

Text values are called **strings** (because they're a *string* of characters).
Often we want to build a sentence that includes a number we calculated. The
easiest way is an **f-string**: put the letter `f` right before the opening
quote, then wrap any variable in curly braces `{ }`.

```python
sample_name = "DRUG3"
density = 62.5

print(f"Sample {sample_name} had a density of {density} cells per mm^2.")
```

You can even format the number inside the braces. Writing `{density:.1f}` means
"show this number with 1 digit after the decimal point" — handy for tidy
reports.

```python
messy_number = 62.487193
print(f"Rounded neatly: {messy_number:.1f}")   # one decimal place
print(f"Rounded neatly: {messy_number:.2f}")   # two decimal places
```

## Step 3 — Lists and loops

### 3a. Lists: holding several values at once

So far each variable held a single value. A **list** holds many values in
order, inside square brackets `[ ]`, separated by commas. This is how we'll
store, say, the density measured in each of five control samples.

```python
con_densities = [48.2, 51.7, 46.9, 53.1, 49.5]   # five measurements

print(con_densities)
print("Number of values in the list:", len(con_densities))   # len() counts them
```

We can ask Python simple questions about a list without doing the arithmetic
ourselves. `sum()` adds everything up; dividing by `len()` gives the average
(the mean).

```python
total = sum(con_densities)
count = len(con_densities)
mean = total / count

print(f"Total:   {total}")
print(f"Count:   {count}")
print(f"Mean:    {mean:.2f} cells per mm^2")
```

### 3b. Pulling one value out of a list (indexing)

A list keeps its values in order, and each item has a position number called an
**index**. Here's the one surprise for beginners: Python starts counting at
**0**, not 1. So the *first* item is index `0`, the *second* is index `1`, and
so on. You grab an item by writing its index in square brackets right after the
list's name.

```python
con_densities = [48.2, 51.7, 46.9, 53.1, 49.5]

print(con_densities[0])    # the FIRST value  (index 0)
print(con_densities[1])    # the SECOND value (index 1)
print(con_densities[-1])   # a handy shortcut: -1 means the LAST value
```

### 3c. Repeating an action with a `for` loop

Copying the same line five times is tedious and easy to get wrong. A
**`for` loop** runs the same code once for *every* item in a list. Read it as a
plain sentence: *"**for** each value **in** this list, do the indented block
below."*

Two things to notice in the code:
- the `for` line ends with a **colon `:`**
- the repeated line is **indented** (pushed in). Python uses that indentation to
  know what belongs *inside* the loop.

The word `value` is just a name *we* chose for "the current item" — Python fills
it in with each list element in turn.

```python
con_densities = [48.2, 51.7, 46.9, 53.1, 49.5]

for value in con_densities:
    print(f"This sample had a density of {value} cells per mm^2")
```

> **⚙️ Try it:** How many lines did that loop print? (It runs once per item in
> the list.) Add a sixth number to `con_densities`, re-run, and confirm the
> number of printed lines changes to match.

Loops can also **build something up** as they go. Here we start a counter at 0
and add 1 every time we meet a sample above a threshold. This also sneaks in an
**`if`** — a decision whose indented line only runs when the condition is true.
You'll see both loops and `if` throughout the later units.

```python
con_densities = [48.2, 51.7, 46.9, 53.1, 49.5]
count_high = 0

for value in con_densities:
    if value > 50:
        count_high = count_high + 1

print(f"{count_high} samples were above 50 cells per mm^2")
```

## Step 4 — Functions and libraries

### 4a. Writing your own function with `def`

You've been *using* functions other people wrote (`print`, `len`, and soon
`np.mean`). You can also write your **own** with the keyword `def`. A function
packages up a calculation, gives it a name, and lets you reuse it with different
inputs. The value it hands back is marked with `return`.

Why learn this now? In the later units you'll constantly see small custom
functions (for example, one that computes an interquartile range). Recognizing
the **`def ... return`** pattern means you'll be able to read that code without
getting stuck.

```python
def cells_per_mm2(cells, area):
    density = cells / area
    return density

# now we can 'call' our function with different numbers:
print(cells_per_mm2(250, 4.0))
print(cells_per_mm2(500, 4.0))
```

`cells` and `area` are placeholders (called **parameters**) — whatever numbers
you pass in get slotted into them when the function runs.

### 4b. Libraries: borrowing tools other people built

Python comes with some built-in commands (like `print`, `sum`, `len`), but the
really powerful tools live in **libraries** — collections of ready-made
functions we can borrow. To use one, we **import** it, usually giving it a
short nickname.

Three libraries you'll see in every future unit:

- **numpy** (nickname `np`) — fast math and statistics
- **pandas** (nickname `pd`) — spreadsheet-style data tables
- **matplotlib** (nickname `plt`) — plots and charts

In Google Colab these are already installed, so we can import them directly.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

Now watch how much shorter the mean calculation becomes when we let `numpy` do
the work. It even gives us the standard deviation (a measure of spread) for
free — you'll learn exactly what that means in the next notebook.

```python
con_densities = [48.2, 51.7, 46.9, 53.1, 49.5]

print(f"Mean (numpy):     {np.mean(con_densities):.2f}")
print(f"Std dev (numpy):  {np.std(con_densities, ddof=1):.2f}")
```

## Step 5 — Data tables, plots, and error messages

### 5a. A first look at a DataFrame (a data table)

Everything so far used plain lists. But real experimental data is a **table** —
rows and columns, like a spreadsheet. The pandas library (`pd`) gives us a table
object called a **DataFrame**. In Unit 1 you'll load one straight from a file;
here we'll build a tiny one by hand just to see what it looks like.

```python
data = {
    "Group":   ["CON", "CON", "DRUG", "DRUG"],
    "Density": [48.2, 51.7, 63.4, 66.1],
}
df = pd.DataFrame(data)
df    # writing df on its own line displays the table nicely in a notebook
```

That `{ ... }` is a **dictionary** — pairs of a label and its values, which here
become the column names. (Dictionaries show up now and then in the later units,
so it's worth seeing one.)

**Selecting a column** works just like indexing a list, except you use the
column's **name** in the brackets. This one move is exactly how Unit 1 begins.

```python
print(df["Density"])                        # pull out just the Density column
print()
print("Mean density:", df["Density"].mean())  # a DataFrame column can do stats too

# You can also FILTER rows by a condition — you'll see this a lot in Units 1 and 3:
con_only = df[df["Group"] == "CON"]
print()
print("Just the CON rows:")
print(con_only)
```

### 5c. Don't panic: reading error messages

Everyone — including experienced scientists — gets **error messages** all the
time. A red error box does **not** mean you broke the computer. It's just
Python telling you it got confused and pointing at the line where it happened.

The cell below has a deliberate mistake: a variable named `speed` is used but
was never created. Run it and look at the last line of the red message — it
usually tells you exactly what went wrong.

```python
# This cell is SUPPOSED to fail — that's the lesson!
print(speed)   # 'speed' was never defined, so Python doesn't know what it is
```

The last line probably said something like `NameError: name 'speed' is not
defined`. In plain English: *"You asked me to use `speed`, but you never told
me what `speed` is."*

**Fix it:** In the cell below we define `speed` first, so the same `print`
works. Run it to confirm.

```python
speed = 10
print(speed)   # now Python knows what 'speed' is
```

The three errors you'll meet most often this term:

- **`NameError`** — you used a name that doesn't exist yet (often a typo, or you
  forgot to run the cell above where it was created).
- **`SyntaxError`** — a punctuation problem, like a missing `)` or quotation mark.
- **`FileNotFoundError`** — Python couldn't find a file (usually the path is
  misspelled). You'll see this once we start loading data files.

The golden rule: **read the very last line of the error**, and check that you
ran every cell above in order.
